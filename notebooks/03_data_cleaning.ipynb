Runs cleaning scripts for all three sources. Raw data is never overwritten.
Cleaning always reads from staging and writes to processed.

> **Scope:** Data engineering only.

```python
import subprocess, sys
def run_script(path):
    r = subprocess.run([sys.executable, path], capture_output=True, text=True, cwd='..')
    print(f'=== {path} (exit {r.returncode}) ===')
    print(r.stdout)
    if r.returncode != 0: print('STDERR:', r.stderr)
```

```python
run_script('../scripts/clean_imdb.py')
```

```python
run_script('../scripts/clean_critics.py')
```

```python
run_script('../scripts/clean_reddit.py')
```
