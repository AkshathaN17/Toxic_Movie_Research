# 02 — Data Collection

**Film:** Toxic: A Fairytale for Grown-Ups | tt27530512 | TOXIC_2026

This notebook orchestrates data collection for all three sources.
It calls the same scripts that can be run from the command line.

**Run order:** IMDb → Critics → Reddit

> **Scope:** Data engineering only. Raw data is preserved after collection.

## A. IMDb Non-Commercial Datasets

Downloads and extracts the TOXIC_2026 record. Requires network access.

```python
import subprocess, sys
result = subprocess.run([sys.executable, '../scripts/download_imdb.py'], capture_output=True, text=True, cwd='..')
print(result.stdout)
if result.returncode != 0:
    print('STDERR:', result.stderr)
```

## B. Critics — Manual Entry

Creates the entry template. Fill it manually then re-run to stage.

```python
result = subprocess.run([sys.executable, '../scripts/collect_critics.py'], capture_output=True, text=True, cwd='..')
print(result.stdout)
if result.returncode != 0:
    print('STDERR:', result.stderr)
```

## C. Reddit — PRAW OAuth

Requires credentials in `.env` and `reddit.enabled: true` in `config/settings.yaml`.

```python
result = subprocess.run([sys.executable, '../scripts/collect_reddit.py', '--dry-run'], capture_output=True, text=True, cwd='..')
print(result.stdout)
print('Exit code:', result.returncode)
# If dry-run passes, remove --dry-run to collect:
```

## D. Collection Log

Shows all collection operations recorded in `logs/collection_log.csv`.

```python
import pandas as pd
from pathlib import Path
log_path = Path('../logs/collection_log.csv')
if log_path.exists():
    display(pd.read_csv(log_path))
else:
    print('No collection log yet — run scripts above first.')
```
