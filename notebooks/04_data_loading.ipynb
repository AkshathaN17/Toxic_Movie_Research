```python
import subprocess, sys
def run_script(path, *args):
    r = subprocess.run([sys.executable, path, *args], capture_output=True, text=True, cwd='..')
    print(f'=== {path} (exit {r.returncode}) ===')
    print(r.stdout)
    if r.returncode != 0: print('STDERR:', r.stderr)
```
